```python
import tensorflow as tf
from tensorflow import keras

print('Good imports!')
```

```python
model_deep = keras.models.Sequential([
    keras.layers.Flatten(input_shape=[28,28]),

    keras.layers.Dense(300, activation='relu', kernel_initializer='he_normal'),
    keras.layers.Dense(100, activation='relu', kernel_initializer='he_normal'),

    keras.layers.Dense(10, activation='softmax')
])
print('Модель успешно создана')
model_deep.summary()

weights, biases = model_deep.layers[1].get_weights()
print(f"\nСреднее значение весов He: {weights.mean():.4f}")
print(f"Стандартное отклонение весов He: {weights.std():.4f}")
```
