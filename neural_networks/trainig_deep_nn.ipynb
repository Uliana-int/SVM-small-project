```python
import tensorflow as tf
from tensorflow import keras

print('Good imports!')
```

```python
model_deep = keras.models.Sequential([
    keras.layers.Flatten(input_shape=[28,28]),

    keras.layers.Dense(300, activation='relu', kernel_initializer='he_normal'),
    keras.layers.Dense(100, activation='relu', kernel_initializer='he_normal'),

    keras.layers.Dense(10, activation='softmax')
])
print('Модель успешно создана')
model_deep.summary()

weights, biases = model_deep.layers[1].get_weights()
print(f"\nСреднее значение весов He: {weights.mean():.4f}")
print(f"Стандартное отклонение весов He: {weights.std():.4f}")
```

```python
model_leaky = keras.models.Sequential([
    keras.layers.Input(shape=(28, 28)),
    keras.layers.Flatten(),
    keras.layers.Dense(100, kernel_initializer='he_normal'),
    keras.layers.LeakyReLU(negative_slope=0.01),

    keras.layers.Dense(10, activation='softmax')
])

print('Модель успешно собрана!')

model_selu = keras.models.Sequential([
    keras.layers.Input(shape=(28, 28)),
    keras.layers.Flatten(),

    keras.layers.Dense(100, activation="selu", kernel_initializer="lecun_normal"),
    keras.layers.Dense(100, activation="selu", kernel_initializer="lecun_normal"),
    keras.layers.Dense(10, activation="softmax")
])

print('Самонормализующая модель успешно собрана!')
```

```python
model_bn = keras.models.Sequential([
    keras.layers.Input(shape=(28,28)),
    keras.layers.Flatten(),

    keras.layers.Dense(300, use_bias=False),
    keras.layers.BatchNormalization(),
    keras.layers.Activation('relu'),

    keras.layers.Dense(100, use_bias=False),
    keras.layers.BatchNormalization(),
    keras.layers.Activation('relu'),

    keras.layers.Dense(10, activation='softmax')
])

print('Архитектура с пакетной нормализацией успешно создана!')
model_bn.summary()
```

```python
optimizer_clipped = keras.optimizers.SGD(clipvalue=1.0)

model_bn.compile(loss='sparse_categorical_crossentropy',
                 optimizer=optimizer_clipped,
                 metrics=['accuracy']
                )

print('Модель успешно скомпилирована!')
```
